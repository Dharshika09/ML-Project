```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# loading the csv data to a Pandas DataFrame
heart_data = pd.read_csv('/content/heart_disease_data.csv')
```

1 --> Defective Heart

0 --> Healthy Heart

Splitting the Features and Target

```python
X = heart_data.drop(columns='target', axis=1)
Y = heart_data['target']
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training



## Logistic Regression

```python
model = LogisticRegression()
```

```python
# training the LogisticRegression model with Training data
model.fit(X_train, Y_train)
```

```python
# accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on Training data : ', training_data_accuracy)
```

```python
# accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy on Test data : ', test_data_accuracy)
```

Building a Predictive System

```python
input_data  = (62,0,0,140,268,0,0,160,0,3.6,0,2,2)

#change the input data to a numpy array
input_data_as_numpy_array= np.asarray(input_data)

#reshape the numpy array as we are predicting for only on instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = model.predict(input_data_reshaped)
print(prediction)


# Step 6: Output result
if prediction[0] == 0:
    print('The Person does not have a Heart Disease')
else:
    print('The Person has Heart Disease')
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

## 🌳 Decision Tree

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train, Y_train)

dt_train_pred = dt_model.predict(X_train)
dt_test_pred = dt_model.predict(X_test)

print('Decision Tree Training Accuracy:', accuracy_score(Y_train, dt_train_pred))
print('Decision Tree Test Accuracy:', accuracy_score(Y_test, dt_test_pred))
```

## 🌲 Random Forest

```python
rf_model = RandomForestClassifier()
rf_model.fit(X_train, Y_train)

rf_train_pred = rf_model.predict(X_train)
rf_test_pred = rf_model.predict(X_test)

print('Random Forest Training Accuracy:', accuracy_score(Y_train, rf_train_pred))
print('Random Forest Test Accuracy:', accuracy_score(Y_test, rf_test_pred))
```

##📏 Support Vector Machine (SVM)


```python
svm_model = SVC()
svm_model.fit(X_train, Y_train)

svm_train_pred = svm_model.predict(X_train)
svm_test_pred = svm_model.predict(X_test)

print('SVM Training Accuracy:', accuracy_score(Y_train, svm_train_pred))
print('SVM Test Accuracy:', accuracy_score(Y_test, svm_test_pred))
```

## 👥 K-Nearest Neighbors (KNN)

```python
knn_model = KNeighborsClassifier()
knn_model.fit(X_train, Y_train)

knn_train_pred = knn_model.predict(X_train)
knn_test_pred = knn_model.predict(X_test)

print('KNN Training Accuracy:', accuracy_score(Y_train, knn_train_pred))
print('KNN Test Accuracy:', accuracy_score(Y_test, knn_test_pred))
```
